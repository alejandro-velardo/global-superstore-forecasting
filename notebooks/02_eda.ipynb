```python
%load_ext autoreload
%autoreload 2

import numpy as np
import pandas as pd

from superstore import config
from superstore.viz.plots import save_fig, set_style

set_style()
pd.set_option("display.max_columns", 50)
```

```python
from superstore.features.build import aggregate_panel
from superstore.viz.plots import plot_discount_profit, plot_series

orders = pd.read_parquet(config.ORDERS_FILE)
monthly = aggregate_panel(orders, freq="MS", group_cols=["category"])
```

## Trend & seasonality

```python
ax = plot_series(monthly, y="sales", hue="category")
save_fig(ax.figure, "sales_monthly_by_category")
```

```python
# Seasonal decomposition (statsmodels) of total monthly sales
from statsmodels.tsa.seasonal import STL

total = monthly.groupby("period")["sales"].sum().asfreq("MS")
STL(np.log1p(total), period=12).fit().plot();
```

## Discount impact on profitability

```python
ax = plot_discount_profit(orders, by="category")
save_fig(ax.figure, "margin_by_discount_band")
```

## Ideas
- Loss-making share by sub-category × discount band (heatmap)
- Market × month heatmap of sales (seasonality stability)
- Profit vs discount scatter with LOWESS per category (non-linearity / threshold)
- Pareto: % of profit from top customers/products
- ACF/PACF of monthly sales → informs lag choice
